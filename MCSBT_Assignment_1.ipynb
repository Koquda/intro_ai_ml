## Read the provided cb.csv into Colab with multiple columns into a list or array of dictionaries, each row is a dictionary. Note: There are repeated rows.

```python
import csv

with open("cb_sample.csv", mode="r", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

print(rows)
```

## Sort the dictionary by the "rank" field in ascending order:
- Copying values into an array or list, and sorting, or
- Using Python sorted function on the dictionary


```python

def get_rank(row):
  return int(row['rank'])

sorted_rows = sorted(rows, key=get_rank)

for row in sorted_rows:
    print(f"Row:{row['name']}, Rank: {row['rank']}")
```

## Create a set containing only unique (non-repeated) rows from the list of dictionaries

```python
# Creating set
unique_set = set()

for row in sorted_rows:
  unique_set.add(tuple(row.items()))

for row in unique_set:
  print(row)
```
