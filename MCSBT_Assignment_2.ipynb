```python
# location of the downloaded binary
filepath = 'drive/MyDrive/Colab Notebooks/intro_ai_ml/assignment2/GoogleNews-vectors-negative300.bin.gz'
```

```python
model = KeyedVectors.load_word2vec_format(filepath, binary=True)
```

## Load amazon dataset

```python
import json
import re
from pathlib import Path

dataset_path = "/content/drive/MyDrive/Colab Notebooks/intro_ai_ml/assignment2/webzio_amazon.json"

articles = []

with open(dataset_path, "r", encoding="utf-8") as file:
    for line in file:
        if line.strip():
            articles.append(json.loads(line))

print("Articles loaded:", len(articles))
```

## Remove duplicates

```python
titles = []
seen_titles = set()

for article in articles:
    title = article.get("title", "").strip()

    if title and title not in seen_titles:
        titles.append(title)
        seen_titles.add(title)

print("Unique titles:", len(titles))

for index in range(10):
    print(index, "-", titles[index])
```

## Article selection

```python
selected_index = 0
selected_title = titles[selected_index]

selected_words = []

for word in selected_title.split():
    if word in model:
        selected_words.append(word)

if not selected_words:
    raise ValueError("Choose another title: no words were found in the model.")

print("Selected title:", selected_title)
```

## Results

```python
results = []

for title in titles:
    if title == selected_title:
        continue

    title_words = []

    for word in title.split():
        if word in model:
            title_words.append(word)

    if title_words:
        similarity = model.n_similarity(selected_words, title_words)

        results.append({
            "title": title,
            "similarity": float(similarity)
        })


def get_similarity(result):
    return result["similarity"]


results = sorted(results, key=get_similarity, reverse=True)
top_50 = results[:50]
```

## Print results

```python
print("Selected title:", selected_title)
print("\n50 most similar titles:\n")

for rank, result in enumerate(top_50, start=1):
    print(f"{rank}. Similarity: {result['similarity']:.4f}")
    print(result["title"])
    print()
```
